```python
import numpy as np
import matplotlib.pyplot as plt
```

### Legends

```python
x = np.linspace(0, 11, 10)
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# Plotting multiple plots on one figure
ax.plot(x,x)
ax.plot(x, x**2);
```

```python
# To label each of the lines (plots)
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# Plotting multiple plots on one figure
ax.plot(x,x, label='X vs X')
ax.plot(x, x**2, label='X vs X^2');

ax.legend()
```

```python
# In cases where there are multiple lines and the labels could overlap with the plotted lines, 
# To label each of the lines (plots)
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# Plotting multiple plots on one figure
ax.plot(x,x, label='X vs X')
ax.plot(x, x**2, label='X vs X^2');

# You can use the loc parameter
ax.legend(loc='lower left')
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# Plotting multiple plots on one figure
ax.plot(x,x, label='X vs X')
ax.plot(x, x**2, label='X vs X^2');

# You can use the loc parameter
# Using the numerical values that are associated with it
ax.legend(loc=4)
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# Plotting multiple plots on one figure
ax.plot(x,x, label='X vs X')
ax.plot(x, x**2, label='X vs X^2');

# You can use the loc parameter
# Setting the labels to a manual position
ax.legend(loc=(1.1, 0.5));
```

### Visual Styling (line styles, line widths, markers, colors, etc.)

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
ax.plot(x,x)
```

```python
# Changing the color
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

ax.plot(x, x, color='lightgreen') 

# you can also insert RGB or HEX codes
ax.plot(x, x+1, color='#8a188c')
```

```python
# Adding labels
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

ax.plot(x, x, color='lightgreen', label='x against x') 

# you can also insert RGB or HEX codes
ax.plot(x, x+1, color='#8a188c', label='x against x+1')

ax.legend()
```

```python
# Line and marker styles
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# Line width lw and linewidth
ax.plot(x, x, color='lightgreen', lw=5) 
ax.plot(x, x+1, color='#8a188c', linewidth=0.5)
```

```python
# Changing the line types
# The possible linestyle options are: '--', '-', '-.', ':', 'steps'
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# ls or linestyle
ax.plot(x, x, color='lightgreen', lw=3, ls='--') 
ax.plot(x, x+1, color='#8a188c', linewidth=5, linestyle=':')
```

```python
# We can also set custom line styles
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

lines = ax.plot(x, x+1, color='#8a188c', linewidth=5, linestyle=':')   # lines is a list object
lines[0].set_dashes([1,1,1,2,3,5])  # The order is: line, space, line, space, ...
```

```python
# We can also set custom line styles
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

lines = ax.plot(x, x+1, color='#8a188c', linewidth=5, linestyle=':')   # lines is a list object
lines[0].set_dashes([1, 2, 1, 2, 10, 2])  # The order is: line, space, line, space, ...
```

```python
# Markers (mark each of the points on the plot)
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# There are multiple marker string codes you can use. Check them out on google
ax.plot(x, x+1, color='#8a188c', linewidth=2, linestyle='-.', marker='o')

ax.plot(x, x, color='darkblue', linewidth=1, linestyle='-', marker='P')

# We can change the marker size as well (ms for short)
ax.plot(x, 2*x, color='red', linewidth=1, linestyle='-', marker='1', markersize=20)
```

```python
# Markers (mark each of the points on the plot)
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

# There are other marker parameters such as the marker_face_color, marker_edge_width, marker_edge_color
ax.plot(x, x+1, color='#8a188c', linewidth=2, linestyle='-.', marker='>', ms=15, markerfacecolor='red', markeredgewidth=2, markeredgecolor='orange')
```
