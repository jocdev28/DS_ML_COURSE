```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# DATA
a = np.linspace(0, 10, 11)
b = a ** 4
```

```python
a
```

```python
b
```

```python
x = np.arange(0, 10)
y = 2 * x
```

```python
x
```

```python
y
```

```python
# Create a figure object
fig = plt.figure()

# Define the axes
axes = fig.add_axes([0,0,1,1])

# Plot on those axes
axes.plot(a,b);

# Note: all the above commands should be going into the same cell
```

```python
type(fig)
```

```python
# Adding multiple axes
fig = plt.figure()

# LARGE AXES

axes1 = fig.add_axes([0,0,1,1])
axes1.plot(a,b)

# SMALL AXES
axes2 = fig.add_axes([0.1,0.3,0.5,0.5])
axes2.plot(x,y);
```

```python
# Adding Labels/titles to each set of axes
fig = plt.figure()

# LARGE AXES
axes1 = fig.add_axes([0,0,1,1])
axes1.plot(a,b)

axes1.set_xlim(0,8)
axes1.set_ylim(0,9000)
axes1.set_xlabel('A Axis')
axes1.set_ylabel('B Axis')
axes1.set_title('Outer Plot')

# SMALL AXES
axes2 = fig.add_axes([0.2,0.3,0.5,0.5])
axes2.plot(a,b)

axes2.set_xlim(0,5)
axes2.set_ylim(0,4500)
axes2.set_xlabel('A Axis')
axes2.set_ylabel('B Axis')
axes2.set_title('Zommed In (Inner Plot)');
```

```python
# Figure Parameters
# To change the clarity of the plot, you can use the dpi parameter (dots per inch)
# To change the figure size in inches, you can use the figsize parameter and set the height and width in inches
fig = plt.figure(figsize=(2, 2), dpi=200)

axes1 = fig.add_axes([0,0,1,1])

axes1.plot(a,b)

# Export the figure
# To include the axes and labels, add the bbox_inches parameter and set the value to 'tight'
fig.savefig('figure_with_param.png', bbox_inches='tight');

# Note: If the image is too large, you can change the values of the figure size from the code (here)
```
