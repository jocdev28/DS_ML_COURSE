```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = np.linspace(0, 10, 11)
b = a ** 4
```

```python
x = np.arange(0, 10)
y = 2 * x
```

```python
fig,axes = plt.subplots(nrows=1, ncols=2)
```

```python
# Note: the axes is a numpy object
type(axes)
```

```python
fig,axes = plt.subplots(nrows=1, ncols=2)
axes[0].plot(a,b)
```

```python
fig,axes = plt.subplots(nrows=2, ncols=2)
axes[0][0].plot(a,b)
axes[1][1].plot(x,y)
```

```python
# Loop through an array and plot 
fig,axes = plt.subplots(nrows=3, ncols=1)

for ax in axes:
    ax.plot(x, y)
```

```python
# to space out the plots so that nothing overlaps
fig,axes = plt.subplots(nrows=3, ncols=1)
axes[1].plot(a, b)

# automatically
plt.tight_layout()
```

```python
# to space out the plots so that nothing overlaps
fig,axes = plt.subplots(nrows=2, ncols=2)

axes[0][0].plot(x, y)

axes[0][1].plot(a, b)

axes[1][0].plot(a, b)

axes[1][1].plot(x, y)

fig.subplots_adjust(wspace=0.3, hspace=0.3)
```

```python
# Setting axis and plot titles for individual plots
fig,axes = plt.subplots(nrows=2, ncols=2)

axes[0][0].plot(x, y)

axes[0][1].plot(a, b)
axes[0][1].set_ylabel('Y Axis')
axes[0][1].set_title('Graph of A against B')

axes[1][0].plot(a, b)

axes[1][1].plot(x, y)
axes[1][1].set_xlim(0, 5)

# Setting a title for the entire figure (you can do some alignment using other parameters)
fig.suptitle('Figure Plot')

plt.tight_layout()

# Save this plot
fig.savefig('mysubplots.png', bbox_inches='tight')
```
