```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
x = np.arange(0, 10)
```

```python
y = 2*x
```

```python
plt.plot(x, y)
```

```python
# Display the plot without the object description at the top (you can also add a semi-colon when using jupyter notebook)
plt.plot(x, y)
plt.show()
```

```python
# Add a title at the top of the plot
plt.plot(x, y)
plt.title('Graph of y against x');
```

```python
# Add an 'x' label and a 'y' label for the axis
plt.plot(x, y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph of y against x');
```

```python
plt.plot(x, y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
# Add axis limits
plt.xlim(0, 6)
plt.ylim(0, 15)
plt.title('Graph of y against x');
```

```python
# Saving a plot
plt.plot(x, y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
# Add axis limits
plt.xlim(0, 6)
plt.ylim(0, 15)
plt.title('Graph of y against x');
# Let's save the plot as a png file
plt.savefig('myfirstplot.png')
```
