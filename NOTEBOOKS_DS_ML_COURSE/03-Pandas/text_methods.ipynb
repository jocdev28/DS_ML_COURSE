```python
messy_names = pd.Series(['andrew  ', 'bo;bo', '   claire   '])
```

```python
messy_names
```

```python
messy_names.str.replace(';','')
```

```python
# use the .strip call to eliminate all unecessary white space
messy_names.str.replace(';','').str.strip()
```

```python
messy_names.str.replace(';','').str.strip().str.capitalize()
```

```python
# To make all these editings easier, we can simply use functions and the apply method
def cleanup(name):
    name = name.replace(";","")
    name = name.strip()
    name = name.capitalize()
    return name
```

```python
messy_names.apply(cleanup)
```
