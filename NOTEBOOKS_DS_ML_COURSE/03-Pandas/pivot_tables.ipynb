```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Sales_Funnel_CRM.csv')
```

```python
df
```

```python
# help(pd.pivot)
```

```python
licenses = df[['Company', 'Product', 'Licenses']]
```

```python
licenses
```

```python
pd.pivot(data=licenses, index='Company', columns='Product')
```

```python
# Calling an aggregation table on this
pd.pivot_table(df, index='Company', aggfunc='sum')
```

```python
# We can do the same using the groupby method
df.groupby('Company').sum(numeric_only=True)
```

```python
df.groupby('Company').sum()
```

```python
# We can also specify the columns we're interested in
pd.pivot_table(df, index='Company', aggfunc='sum', values=['Licenses', 'Sale Price'])
```

```python
# Group by account manager and contact, and get the total sale price per contact
pd.pivot_table(df, index=['Account Manager', 'Contact'], values=['Sale Price'], aggfunc='sum')
```

```python
pd.pivot_table(df, index=['Account Manager', 'Contact'], values=['Sale Price'], columns=['Product'], aggfunc='sum')
```

```python
# Add a fill value for missing numbers
pd.pivot_table(df, index=['Account Manager', 'Contact'], values=['Sale Price'], columns=['Product'], aggfunc='sum', fill_value=0)
```

```python
# We can also pass in direct numpy calls
pd.pivot_table(df, index=['Account Manager', 'Contact'], values=['Sale Price'], columns=['Product'], aggfunc=[np.sum], fill_value=0)
```

```python
# We can pass in multiple
pd.pivot_table(df, index=['Account Manager', 'Contact'], values=['Sale Price'], columns=['Product'], aggfunc=[np.sum, np.mean], fill_value=0)
```

```python
# However, instead of setting multiple aggregate functions in one line of code, we can split them into separate ones
pd.pivot_table(df, index=['Account Manager', 'Contact', 'Product'], values=['Sale Price'], aggfunc=[np.sum], fill_value=0)
```

```python
# There is a convenience parameter called 'margins'. It gives a grand total of the values
pd.pivot_table(df, index=['Account Manager', 'Contact', 'Product'], values=['Sale Price'], aggfunc=[np.sum], fill_value=0, margins=True)
```
