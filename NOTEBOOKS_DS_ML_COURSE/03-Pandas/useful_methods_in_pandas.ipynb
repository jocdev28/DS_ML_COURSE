### The .apply() Method

```python
# The apply() method allows you to apply any custom python function to any every row in a series. We can do this using either one or multiple columns.
def last_four(num):
    return str(num)[-4:]
```

```python
last_four(123456789)
```

```python
# Now, applying the function to the data and creating the separate column for the new data
df['last_four'] = df['CC Number'].apply(last_four)
```

```python
df.head()
```

```python
df['total_bill']
```

```python
df['total_bill'].mean()
```

```python
def yelp(price):
    if price < 10:
        return '$'
    elif price >= 10 and price < 30:
        return '$$'
    else:
        return '$$$'
```

```python
df['yelp_review'] = df['total_bill'].apply(yelp)
```

### Applying custom functions to multiple columns

```python
# Lambda expressions: A Lambda expression is a small, anonymous function that is defined without a name and the def keyword.
# Note that not every function can be converted to a lambda expression, like more complex functions
lambda num: num * 2
```

```python
df['total_bill'].apply(lambda num: num * 2)
```

```python
def quality(total_bill, tip):
    if tip/total_bill > 0.25:
        return "Generous"
    else:
        return "Other"
```

```python
df[['total_bill', 'tip']].apply(lambda df: quality(df['total_bill'], df['tip']), axis = 1)
```

```python
# Create another column for this data
df['Quality'] = df[['total_bill', 'tip']].apply(lambda df: quality(df['total_bill'], df['tip']), axis = 1)
```

```python
df.head()
```

```python
# Using the numpy vectorize method
df['Quality'] = np.vectorize(quality)(df['total_bill'], df['tip'])
df.head()
```

#### Getting the index location of the min and max values

```python
# Getting the minimum value:
df['total_bill'].min()
```

```python
# Getting the maximum value:
df['total_bill'].max()
```

```python
# Getting the index value of the maximum value
df['total_bill'].idxmax()
```

```python
# Getting the index value of the minimum value
df['total_bill'].idxmin()
```

```python
# Grab the row that has the minimum total bill
df.iloc[df['total_bill'].idxmin()]
```

```python
# Grab the row that has the maximum total bill
df.iloc[df['total_bill'].idxmax()]
```

iloc means "integer location" and it is used for integer-location based indexing to select specific rows and columns from a DataFrame or Series

##### Mapping is the better option when you want to replace a lot of items while the replace method is best for smaller items
