```python
# Imaginary sales data for 1st and 2nd Quarters for Global Company
q1 = {'Japan': 80, 'China': 450, 'India': 200, 'USA': 250}
q2 = {'Brazil': 100, 'China': 500, 'India':210, 'USA': 260}
```

```python
sales_q1 = pd.Series(q1)
```

```python
sales_q2 = pd.Series(q2)
```

```python
sales_q1
```

```python
sales_q2
```

```python
sales_q1['China']
```

```python
sales_q1[1]
```

```python
# Get all the keys in a Pandas series
sales_q1.keys()
```

### Performing Broadcast operations

```python
# Performing arithmetic operations using braodcast
sales_q1 * 2
```

```python
sales_q1 / 100
```

```python
# Pandas does not know what to do with series that have different elements when adding them up together
# Because Japan and Brazil are not present in both series, it replaces the sum value with NaN
sales_q1 + sales_q2
```

```python
# To Bypass this, we can use the add() function
sales_q1.add(sales_q2, fill_value=0)
```

```python
# Creating a new column
df['tip_percentage'] = 100 * df['tip'] / df['total_bill']
```

```python
df.head()
```

```python
# Overwritting an already existing column
df['price_per_person'] = df['total_bill'] / df['size']
```

```python
df.head()
```

```python
# Round up values using NumPy's round method
df['price_per_person'] = np.round((df['total_bill'] / df['size']), 2)
df['tip_percentage'] = np.round(df['tip_percentage'], 2)
```

```python
df.head()
```

```python
# removing a column or row. To remove a column, we set the axis to 1 and 0 to remove a row
df.drop('tip_percentage', axis=1)    # Note: This does not affect the original DataFrame
```

```python
df
```

```python
# to make the change permanent, we have to include the 'inplace' argument and set it to True
# or we can simply set the df variable to the new value where the column has been removed
df = df.drop('tip_percentage', axis=1)
df
```
