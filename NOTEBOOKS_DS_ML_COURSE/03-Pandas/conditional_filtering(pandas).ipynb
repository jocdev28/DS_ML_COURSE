```python
df[df['smoker'] == 'Yes']
```

```python
# Filtering based on multiple conditions
# Using Pandas' AND and OR operators (& and |)
```

```python
# Checking if two conditions are both true using Pandas and operator (&)
df[(df['tip'] > 5) & (df['size'] > 4)]
```

```python
df[(df['Payment ID'] == 'Sun5260') | (df['Payment ID'] == 'Sat1766')]
```

```python
# To compare multiple options in the same category, we can use the 'isin' function
options = ['Sat', 'Sun']
```

```python
df['day'].isin(options)
```

```python
# Get only the rows where it is True
df[df['day'].isin(options)]
```
