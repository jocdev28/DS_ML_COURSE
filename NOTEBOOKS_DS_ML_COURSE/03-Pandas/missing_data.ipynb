```python
df = pd.read_csv('movie_scores.csv')
```

```python
df
```

```python
# Check for and select null values
df.isnull()
```

```python
# Check for non-null values
df.notnull()
```

```python
# Checking the columns for any missing values in the specific column
df['pre_movie_score'].notnull()
```

```python
# To get back the speific rows that do no have null values in that specific column
df[df['pre_movie_score'].notnull()]
```

```python
# to get back the rows that have missing values in the pre_movie_score column
df[df['pre_movie_score'].isnull()]
```

```python
# Now, for the option to manage missing data:
# Keeping the data is basiclly leaving it as is
```

```python
# Dropping the data usong df.dropna
# help(df.dropna)
```

```python
# The dropna() method drops any rows that have any missing values
df.dropna()
```

```python
# If we only want to drop the row that has all its values missing
# This drops every row that has null values except they have at least 1 non-null value
df.dropna(thresh=1)
```

```python
# Now, to replace the values
help(df.fillna)
```
