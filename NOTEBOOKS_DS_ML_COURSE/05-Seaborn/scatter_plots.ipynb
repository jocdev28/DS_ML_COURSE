```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv('dm_office_sales.csv')
```

```python
df.head()
```

```python
# Creating a scatter plot
sns.scatterplot(x='salary', y='sales', data=df);
```

```python
# Seaborn works on matplotlib, so you can edit your plot using Matplotlib before Seaborn
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df);
```

```python
# Hue adds variations of a color to the plot (we can hue either by a continuous column or a categorical one
# Hue by continuous column
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, hue='salary');
```

```python
# Hue by categorical column
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, hue='level of education');
```

```python
# To change the colors, you would have to choose a palette from matplotlib's color map
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, hue='level of education', palette='Set1');
```

```python
# We can also change the size of the points based on the values
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, size='work experience');
```

```python
# And we can change the general size of the points
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, s=200);
```

```python
# To alleviate the overlap of the points, we can set an 'alpha' value (this changes the transparency)
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, s=200, alpha=0.2);
```

```python
# We can set the markers
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, style='level of education');
```

```python
# Setting both the colors and shapes
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, style='level of education', hue='level of education');
```

```python
# Save the plot
plt.figure(figsize=(12,4))
sns.scatterplot(x='salary', y='sales', data=df, style='level of education', hue='level of education');

plt.savefig('my_first_plot.png')
```
